```python
def newton_forward():
    n = int(input("Enter number of points: "))
    x = []
    y = []
    for i in range(n):
        x.append(float(input(f"Enter x{i}: ")))
        y.append(float(input(f"Enter y{i}: ")))

    h = x[1] - x[0]
    diff = [y[:]]
    for k in range(1, n):
        prev = diff[-1]
        diff.append([prev[i+1] - prev[i] for i in range(len(prev)-1)])

    xp = float(input("Enter x to find y: "))
    t = (xp - x[0]) / h

    result = diff[0][0]
    t_prod = 1
    fact = 1
    for k in range(1, n):
        t_prod *= (t - (k - 1))
        fact *= k
        result += (t_prod / fact) * diff[k][0]

    print("Interpolated value:", format(result, ".4f"))

newton_forward()
```
