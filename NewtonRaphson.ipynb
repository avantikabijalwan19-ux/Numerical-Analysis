```python
def func(x):
    return x**3 - x**2 + 2

def derivative(x):
    return 3*x**2 - 2*x

def newton_raphson(x):
    for i in range(100):
        x = x - func(x) / derivative(x)
        if abs(func(x)) < 0.0001:
            break
    print("Root:", format(x, ".4f"))

x = float(input("Enter initial guess: "))
newton_raphson(x)
```
