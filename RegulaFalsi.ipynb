```python
def func(x):
    return x**3 - 2*x - 5

def regula_falsi(a, b):
    if func(a) * func(b) >= 0:
        print("Wrong a and b")
        return
    c = a
    for i in range(100):
        c = (a * func(b) - b * func(a)) / (func(b) - func(a))
        if func(c) == 0.0:
            break
        if func(c) * func(a) < 0:
            b = c
        else:
            a = c
    print("Root:", format(c, ".4f"))

a = float(input("Enter a: "))
b = float(input("Enter b: "))
regula_falsi(a, b)
```
