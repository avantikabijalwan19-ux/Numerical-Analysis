```python
def gauss_seidel():
    n = int(input("Enter number of equations: "))
    A = []
    b = []
    for i in range(n):
        row = list(map(float, input(f"Enter row {i+1} coefficients: ").split()))
        A.append(row)
        b.append(float(input(f"Enter b{i+1}: ")))

    x = [0.0] * n
    for k in range(100):
        x_old = x[:]
        for i in range(n):
            s = sum(A[i][j] * x[j] for j in range(n) if j != i)
            x[i] = (b[i] - s) / A[i][i]
        if max(abs(x[i] - x_old[i]) for i in range(n)) < 0.0001:
            break

    for i in range(n):
        print(f"x{i+1} =", format(x[i], ".4f"))

gauss_seidel()
```
